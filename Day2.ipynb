```python
100% abstraction
from abc import ABC, abstractmethod

class Test(ABC):
  @abstractmethod
  def m1(self):
    pass

  @abstractmethod
  def m2(self):
    pass

  class MyPython(Test):
    def m1(self):
      print("m1 implementation")
    def m2(self):
      print("m2 implementation")

  m=MyPython()
  m.m1()
  m.m2()
```

```python
fruit_list1 = ['apple','berry','cherry','papaya']
fruit_list2 = fruit_list1
fruit_list3 = fruit_list1[:]
fruit_list2[0]='guava'
fruit_list3[1]="kiwi"

sum=0
for ls in (fruit_list1,fruit_list2,fruit_list3):
  if ls[0]== 'guava':
    sum += 1
  if ls[1] == 'kiwi':
    sum += 20
```

```python
data = [[[1,2],[3,4]],[[5,6],[7,8]]]

def fun(m):
  v = m[0][0]
  for row in m:
    for element in row:
      if v < element:
        v = element

  return v

print(fun(data[0]))
```
