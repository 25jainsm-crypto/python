```python
import numpy as np
import pandas as pd
import matplotlib as plt
```

```python
import pandas as pd

data = {
    'Name': ['A', 'B', 'C', 'D', 'E'],
    'Math': [78, 34, 76, 89, 92],
    'Science': [21, 70, 54, 76, 88],
    'English': [85, 70, 90, 87, 89]
}

df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Math", "Science", "English"]].values
average_marks = np.mean(marks, axis=1)
df["average"] = average_marks
print(df)
```

```python
def grade(average):
    if average >= 90:
      return 'A'
    elif average >= 80:
      return 'B'
    elif average >= 70:
      return 'C'
    elif average >= 60:
      return 'D'
    else:
      return 'F'

# Use lowercase 'a' for 'average' to match your column name
df['Grade'] = df['average'].apply(grade)
df
```

```python
plt.figure()
plt.bar(df["Name"],df["Average"])
plt.xlabel('students')
plt.ylabel('Average student')
plt.title('Average marks of student')
plt.show()
```
