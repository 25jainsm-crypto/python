```python
import numpy as np
students = np.array([
    [21,55,78],
    [60,43,21],
    [12,43,67],
    [32,67,34],
    [54,21,56],
    ])
students
```

```python
avg_marks = np.mean(students,axis=1)
print("average marks:",avg_marks)
```

```python
topper = np.argmax(avg_marks)
print("topper student:",topper)
```

```python
failed_student = np.where(avg_marks<60)
print("failed student:",failed_student[0]+1)
```

```python
#performance report
for i,avg in enumerate(avg_marks):
  status = "pass" if avg> 50 else "fail"
  print(f"student{i+1}: avg = {avg:.2f},{status}")
```

```python
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

plt.plot(x, y)
plt.title('Line plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.grid(True)
plt.show()
```

```python
# Bar chart
months = ['jan', 'feb', 'mar', 'april']
sales = [250, 122, 43, 121]

plt.bar(months, sales, color='pink')
plt.title('monthly sales')
plt.xlabel('month')
plt.ylabel('sales')
plt.show()
```

```python
plt.barh(months,sales,color='purple')
plt.title('monthly sales(horizontal)')
plt.xlabel('sales')
plt.ylabel('month')
plt.show
```

```python

# scatter plot (comparing height vs weight)
# scatter plot (comparing height vs weight)
height = [150, 160, 170, 180]
weight = [56, 67, 34, 75] # Added a 4th value to match the height list

plt.scatter(height, weight, color='red')
plt.title('height vs weight')
plt.xlabel('height(cm)')
plt.ylabel('weight(kg)')
plt.show()





```

```python
# pie plot (market share)
Brands = ['Brand A', 'Brand B', 'Brand C']
shares = [45, 39, 25]
plt.pie(shares, labels=Brands, autopct='%1.1f%%', startangle=90)
plt.title('market shares')
plt.axis('equal')
plt.show()
```

```python
# multiple plot in one figure
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

plt.plot(x, y1, label='square')
plt.plot(x, y2, label='linear')
plt.legend()
plt.title('multiple lines')
plt.show()
```

```python
# subplots
fig, axs = plt.subplots(2, 2)

axs[0,0].plot(x, y1)
axs[0,0].set_title('square')

axs[0,1].bar(months, sales)
axs[0,1].set_title('sales')

axs[1,0].scatter(height, weight)
axs[1,0].set_title('scatter')

axs[1,1].pie(shares, labels=Brands, autopct='%1.1f%%')
axs[1,1].set_title('pie')

plt.tight_layout()
plt.show()
```
